# Merg

```python
df1 = pd.DataFrame({ 'ID' : [1,2,34,4],
                     'Class' : [5,6,7,8]})
df1
```

```python
df2 = pd.DataFrame({ 'ID' : [1,2,3,4],
                     "Name" : ['A','B','C','D']})
df2
```

```python
pd.merge(df1,df2)
```

```python
pd.merge(df1,df2, on = "ID")
```

```python
pd.merge(df2,df1, on = "ID")
```

```python
# intersection = inner , same as merge on id(same valuse merge different value not)
pd.merge(df1,df2, on = 'ID' , how = 'inner')
```

```python
pd.merge(df1,df2, on = 'ID', how = 'left')
```

```python
pd.merge(df1,df2, on = 'ID', how = 'right')
```

```python
pd.merge(df1,df2, on = 'ID', how = 'outer')
```

```python
pd.merge(df1,df2, on = 'ID', how = 'outer' , indicator  = True)
```

```python
pd.merge(df1,df2, left_index = True, right_index = True)
```

```python
pd.merge(df1,df2, on = 'ID', suffixes = ('_Higher', '_Middle')) # to show get same columns try on class
```
