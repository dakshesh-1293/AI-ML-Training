```python
ml_stud_age = np.random.randint(18,50,(100))
py_stud_age = np.random.randint(18,50,(100))

bins = [15,20,25,30,35,40,45]


plt.hist(ml_stud_age,
         bins,
         histtype= 'barstacked',
         color = 'y',
         label='Ml_stud'
         )

plt.hist(py_stud_age,
         bins,
         histtype= 'barstacked',
         color = 'r',
         label='Py_stud')

plt.title('ML Students Age')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.legend() #we can put this labal in hist perameter also

plt.show()
```

```python
ml_stud_age = np.random.randint(18,50,(100))
py_stud_age = np.random.randint(18,50,(100))

bins = [15,20,25,30,35,40,45]


plt.hist(ml_stud_age,
         bins,
         histtype= 'barstacked',
         rwidth=0.9,
         color = 'y',
         label='Ml_stud',
         orientation='horizontal'
         )

plt.hist(py_stud_age,
         bins,
         histtype= 'barstacked',
         rwidth=0.9,
         color = 'r',
         label='Py_stud')

plt.title('ML Students Age')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.legend() #we can put this labal in hist perameter also

plt.show()
```

```python
ml_stud_age = np.random.randint(18,50,(100))
py_stud_age = np.random.randint(18,50,(100))

bins = [15,20,25,30,35,40,45]


plt.hist([ml_stud_age,py_stud_age,],
         bins,
         histtype= 'bar',
         rwidth=0.9,
         color = ['y','r'],
         label=['Ml_stud','Py_stud'],
         )

plt.title('ML Students Age')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.legend() #we can put this labal in hist perameter also

plt.show()
```
