```python
df = pd.read_csv("googleplaystore.csv", nrows=1000)
df
```

```python
x =df['Rating']
y = df['Reviews']
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```

```python
plt.scatter(x,y)

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.scatter(x,y, color = 'r')

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.scatter(x,y, c = 'r', marker = '*')

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.scatter(x,y, c = 'r', marker = '*', s = 100 , alpha=None) #s for marker size | alpha for

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.scatter(x,y, c = 'r', marker = '*', s = 100 , alpha=0.2) # alpha --> 0 to 1
#s for marker size | alpha for density of points

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(10,7))

plt.scatter(x,y, c = 'r', marker = '*', s = 60 , alpha=None, linewidths=5, edgecolors='b')
plt.scatter(x,df['Installs'], c = 'y', marker = 'o', s = 100 , alpha=0.2, linewidths=5, edgecolors='c')

plt.xlabel("Rating")
plt.ylabel("Installs")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```
