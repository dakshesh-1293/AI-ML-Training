```python
classes = ['Python', 'R', 'AI', 'ML', 'DS']
class1_stud = [30,10,20,25,10]
class2_stud = [4,5,20,20,10]
class3_stud = [3,5,30,15,15]
```

```python
style.use('ggplot')
plt.figure(figsize=(8,4.5))

plt.bar(classes, class1_stud, width = 0.5, align = 'edge', color = 'y',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 1 students')

plt.bar(classes, class2_stud, width = 0.5, align = 'edge', color = 'm',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 2 students')

plt.title('Bar chart for classes', fontsize=15)
plt.xlabel('Classes', fontsize=15)
plt.ylabel('Number of students', fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(8,4.5))

#for putting second clase  side way
classes_index = np.arange(len(classes))

width = 0.3
plt.bar(classes_index, class1_stud, width, align = 'edge', color = 'y',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 1 students')

plt.bar(classes_index + width, class2_stud, width, align = 'edge', color = 'm',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 2 students')

plt.title('Bar chart for classes', fontsize=15)
plt.xlabel('Classes', fontsize=15)
plt.ylabel('Number of students', fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(8,4.5))

#for putting second clase  side way
classes_index = np.arange(len(classes))


plt.bar(classes_index, class1_stud, width=0.2, align = 'edge', color = 'y',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 1 students')

plt.bar(classes_index + 0.3, class2_stud, width=0.2, align = 'edge', color = 'm',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 2 students')

plt.title('Bar chart for classes', fontsize=15)
plt.xlabel('Classes', fontsize=15)
plt.ylabel('Number of students', fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(8,4.5))

#for putting second clase  side way
classes_index = np.arange(len(classes))

width = 0.2
plt.bar(classes_index, class1_stud, width, align = 'edge', color = 'y',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 1 students')

plt.bar(classes_index + width, class2_stud, width, align = 'edge', color = 'm',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 2 students')

plt.xticks(classes_index + width, classes, rotation = 20) #label rotation

plt.title('Bar chart for classes', fontsize=15)
plt.xlabel('Classes', fontsize=15)
plt.ylabel('Number of students', fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(8,4.5))

#for putting second clase  side way
classes_index = np.arange(len(classes))

width = 0.2
plt.bar(classes_index, class1_stud, width, align = 'edge', color = 'y',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 1 students')

plt.bar(classes_index + width, class2_stud, width, align = 'edge', color = 'm',
         edgecolor = 'b', linewidth = 3, alpha = 0.5, linestyle = '--',
        label = 'class 2 students')

plt.xticks(classes_index + width, classes, rotation = 90) #label rotation

plt.title('Bar chart for classes', fontsize=15)
plt.xlabel('Classes', fontsize=15)
plt.ylabel('Number of students', fontsize=15)
plt.show()
```

```python
style.use('ggplot')
plt.figure(figsize=(8,4.5))

#for putting second clase  side way
classes_index = np.arange(len(classes))

width = 0.2
plt.bar(classes_index, class1_stud, width, align = 'edge', color = 'y',
         edgecolor = 'black', linewidth = 2, alpha = 0.5,
        label = 'class 1 students')

plt.bar(classes_index + width, class2_stud, width, align = 'edge', color = 'm',
         edgecolor = 'black', linewidth = 2, alpha = 0.5,
        label = 'class 2 students')

plt.bar(classes_index + width + width, class3_stud, width, align = 'edge', color = 'g',
         edgecolor = 'black', linewidth = 2, alpha = 0.5,
        label = 'class 3 students')

plt.xticks(classes_index + width, classes, rotation = 20) #label rotation

plt.title('Bar chart for classes', fontsize=15)
plt.xlabel('Classes', fontsize=15)
plt.ylabel('Number of students', fontsize=15)
plt.legend()
plt.show()
```
